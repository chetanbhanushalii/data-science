```python
# Write a function in python that can reverse a string using stack data structure. 
# Use Stack class from the tutorial.
# reverse_string("We will conquere COVID-19") should return "91-DIVOC ereuqnoc lliw eW"
```

```python
from collections import deque
stack = deque()
```

```python
def reverse_string(input_str):
    reverse_str = ''
    for ele in input_str[::-1]:
        stack.append(ele)
        reverse_str = reverse_str + stack.pop()
    print(reverse_str)
```

```python
reverse_string("We will conquere COVID-19")
```

```python
# Write a function in python that checks if paranthesis in the string are balanced or not. 
# Possible parantheses are "{}',"()" or "[]". Use Stack class from the tutorial.

# is_balanced("({a+b})")     --> True
# is_balanced("))((a+b}{")   --> False
# is_balanced("((a+b))")     --> True
# is_balanced("))")          --> False
# is_balanced("[a+b]*(x+2y)*{gg+kk}") --> True
```

```python
stack2 = deque()
```

```python
def is_balanced(inp_str):
    open_brackets = ['(','{','[']
    closed_brackets = [')','}',']']
    for ele in inp_str:

        if ele in open_brackets:
            stack2.append(ele)

        if ele in closed_brackets and len(stack2)==0:
            return False
        elif ele in closed_brackets:
            stack2.pop()
        
    
    if len(stack2)==0:
        return True
    else:
        return False
        
```

```python
is_balanced("({a+b})")
```

```python
is_balanced("))((a+b}{")
```

```python
is_balanced("((a+b))") 
```

```python
is_balanced("))")
```
